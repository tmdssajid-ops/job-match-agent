4 jobs with title, city and remote flags. Print only the remote ones.

1. Create a list with the 4 jobs; each job as a dictionary.
2. Use for loop to traverse through the list
3. If remote == True, print only job names.





```python
jobs = [
    {"title": "Software Engineer", "city": "Mumbai", "remote": True},
    {"title": "Data Analyst", "city": "Bengaluru", "remote": False},
    {"title": "Data Scientist", "city": "Pune", "remote": True},
    {"title": "Product Manager", "city": "Delhi", "remote": False}
]

# Traverse and print only remote job titles
for job in jobs:
    if job["remote"] == True:
        print(job["title"])
```

```python
def get_remote_jobs(jobs):
  remote=[]
  for job in jobs:
    if job["remote"]:
      remote.append(job)
  return remote

def get_jobs_in_city(jobs,city):
  city_jobs=[]
  for job in jobs:
    if job["city"] == city:
      city_jobs.append(job)
  return city_jobs

print(get_remote_jobs(jobs))
print(get_jobs_in_city(jobs,"Delhi"))
```
