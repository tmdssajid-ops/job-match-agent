```python
from google import genai
from google.colab import userdata

client = genai.Client(api_key = userdata.get("GEMINI_API_KEY"))
MODEL="gemini-3.5-flash"
response = client.models.generate_content(
    model=MODEL,
    contents="Explain what an API is in one sentence."
)
print(response.text)
```

```python
def ask(prompt):
  response = client.models.generate_content(model="gemini-3.5-flash",contents=prompt)
  return response.text
ask("How do i prepare for a job interview")
```

```python
from pydantic import BaseModel

class JobInfo(BaseModel):
    skills: list[str]
    seniority: str
    remote: bool

description="Your role is a software engineer. Expected skills are python, sql, git and github. This is junior role opening."
response = client.models.generate_content(
    model=MODEL,
    contents="Extract the job info: " + description,
    config={
        "response_mime_type": "application/json",
        "response_schema": JobInfo
    }
)
print(response.parsed)
```

```python
def extract_job_info(description):
  response = client.models.generate_content(
    model=MODEL,
    contents="Extract the job info: " + description,
    config={
        "response_mime_type": "application/json",
        "response_schema": JobInfo
    }
)
  return response.parsed

job_descriptions =["We are looking for a Senior Data Analyst to lead our business intelligence initiatives.The ideal candidate must possess advanced skills in SQL, Tableau, and Python scripting. This is a senior-level position requiring at least 5 years of industry experience.", "Join our engineering team as an Associate Frontend Developer to build interactive user interfaces.You will need hands-on experience with HTML, CSS, JavaScript, and React framework. This is an entry-level/junior role suited for recent graduates or self-taught developers.", "Seeking a Mid-Level DevOps Engineer to automate and streamline our cloud infrastructure operations.Key requirements include proficiency with Docker, Kubernetes, AWS, and CI/CD pipelines. This is a mid-level role for professionals looking to expand their cloud-native expertise."]
for description in job_descriptions:
  extracted_info = extract_job_info(description)
  print(f"Extracted Skills: {extracted_info.skills}")
  print()
```
