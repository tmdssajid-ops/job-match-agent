```python
class Job:
   def __init__(self,title,city):
      self.title= title
      self.city=city
   def describe(self):
      return(f"{self.title} in {self.city}")

job = Job("SW","Pune")
print(job.describe())
```

```python
jobs = [
    {
        "title": "Software Engineer",
        "salary": 120000,
        "city": "San Francisco",
        "skills": ["Python", "SQL", "Docker", "AWS"]
    },
    {
        "title": "Data Analyst",
        "salary": 85000,
        "city": "New York",
        "skills": ["Excel", "Tableau", "Python", "SQL"]
    },
    {
        "title": "Product Manager",
        "salary": 130000,
        "city": "Seattle",
        "skills": ["Agile", "Roadmapping", "SQL"]
    },
    {
        "title": "Data Scientist",
        "salary": 140000,
        "city": "Boston",
        "skills": ["Python", "Machine Learning", "SQL"]
    },
    {
        "title": "Cloud Architect",
        "salary": 155000,
        "city": "Austin",
        "skills": ["AWS", "Kubernetes", "Docker"]
    }
]

count = {}
for job in jobs:
   for skill in job["skills"]:
      count[skill] = count.get(skill, 0) + 1
print("Skill frequency count:", count)

count_sorted = sorted(count.items())
print("Alphabetically sorted skills:", count_sorted)
```

```python
print(sorted(count.items(), key=lambda pair: pair[1], reverse=True)[:3])
```

```python
def is_palindrome(text):
   return text.lower()[::-1]==text.lower()


print(is_palindrome("Level"))
```

```python
def has_duplicates(numbers):
   return len(numbers) != len(set(numbers))

print(has_duplicates([1,2,3,1]))
```
