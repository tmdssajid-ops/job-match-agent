I have a list of jobs. I want to save them to jobs.json, load them back, and print the titles of jobs in a given city. Some jobs might be missing a "city" field.

1. Create a list of dicitonaries. Each dictionary being one job.
2. Open jobs.json in append mode
3. Use for loop to traverse and dump each job in jobs.json
4. Open jobs.json in read mode
5. Use for loop to print the jobs in a given city.
6. Use get() to handle error in case of missing city field

```python
import json
jobs =[
    {'title':"Software engineer","city":"Delhi","remote":False},
    {'title':"Hardware engineer","city":"Noida","remote":True},
    {'title':"Product Manager","remote":True},
    {'title':"Data Analyst","city":"Pune","remote":True},
    {'title':"Operations Manager","city":"Bengaluru","remote":True}
    ]
with open ("jobs.json", "w") as f:
    json.dump(jobs, f, indent = 2)

with open ("jobs.json") as f:
  loaded=json.load(f)
print(f"Number of jobs available : {len(loaded)}")
```

```python
def get_jobs_in_city(jobs,city):
  city_jobs=[]
  for job in jobs:
    if job.get("city") == city:
      city_jobs.append(job)
  return city_jobs
get_jobs_in_city(loaded,"Delhi")
```

```python
def get_jobs_in_city(jobs,city):
  city_jobs=[job for job in jobs if job.get("city")==city]
  return city_jobs
get_jobs_in_city(loaded,"Delhi")
```

```python
def load_jobs(path):
  try:
    with open (path) as f:
      return(json.load(f))
  except FileNotFoundError:
    print("File does not exist. Please check the name again.")
    return []

print(load_jobs("jobs.json"))
```
