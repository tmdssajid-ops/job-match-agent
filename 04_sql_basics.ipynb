```python
import sqlite3
conn = sqlite3.connect("jobs.db")
cur = conn.cursor()
```

```python
cur.execute("CREATE TABLE IF NOT EXISTS jobs (id INTEGER PRIMARY KEY, title TEXT, city TEXT, salary INTEGER)")
cur.execute("INSERT INTO jobs (title, city, salary) VALUES ('SWE Intern', 'Bengaluru', 75000)")
conn.commit()
cur.execute("SELECT * FROM jobs WHERE city = 'Bengaluru'")
print(cur.fetchall())
```

1. All jobs in Pune - WHERE
2. 3 highest paying jobs - ORDER BY, LIMIT
3. How many jobs in each city - GROUP BY, COUNT(title)

```python
# Insert 6 more records to reach a total of 7 records in the table (1 existing + 6 new)
new_jobs = [
    ('Data Analyst', 'Pune', 60000),
    ('Product Manager', 'Bengaluru', 120000),
    ('Software Engineer', 'Pune', 95000),
    ('HR Specialist', 'Mumbai', 55000),
    ('UX Designer', 'Delhi', 80000),
    ('DevOps Engineer', 'Hyderabad', 110000)
]

cur.executemany("INSERT INTO jobs (title, city, salary) VALUES (?, ?, ?)", new_jobs)
conn.commit()

# Verify the insertion by printing all records currently in the table
cur.execute("SELECT * FROM jobs")
print(cur.fetchall())
```

```python
cur.execute("SELECT title FROM jobs WHERE city = 'Pune'")
print(cur.fetchall())
```

```python
cur.execute("SELECT title FROM jobs ORDER BY salary DESC LIMIT 3")
print(cur.fetchall())
```

```python
cur.execute("SELECT city,COUNT(title) FROM jobs GROUP BY city")
print(cur.fetchall())
```

```python
cur.execute("SELECT title FROM jobs WHERE city = 'Bengaluru' AND salary > 50000 ORDER BY salary DESC")
print(cur.fetchall())
```
